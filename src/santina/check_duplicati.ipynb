```python
import pandas as pd

socio = pd.read_csv("dataset_indicatori_definitivo.csv", low_memory=False)

# Tutte le righe la cui chiave (Anno, Codice_NOC) compare più di una volta.
duplicati = (
    socio[socio.duplicated(subset=["Anno", "Codice_NOC"], keep=False)]
    .sort_values(["Anno", "Codice_NOC"])
)

print(f"Righe duplicate: {len(duplicati)} "
      f"({duplicati.groupby(['Anno', 'Codice_NOC']).ngroups} coppie)")

# Colonne essenziali per vedere il pattern: una riga ha le medaglie,
# l'altra ha gli indicatori.
dupl = duplicati[["Anno", "Codice_NOC", "Nazione", "Totale_Medaglie",
           "PIL_Pro_Capite_USD", "Popolazione_Totale", "Aspettativa_di_Vita"]]

dupl["Codice_NOC"].unique()

pd.option_context('display.max_rows', None, 'display.max_columns', None)
duplicati
```
